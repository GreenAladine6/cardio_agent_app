```python
matches = []

for folder in DRIVE_ROOT.rglob("ourdataset/ourdataset"):
    if (
        (folder / "train" / "normal").is_dir()
        and (folder / "train" / "abnormal").is_dir()
    ):
        matches.append(folder)

print("Matching dataset directories:", len(matches))

for folder in matches:
    print(folder)
```

```python
# Replace this with the actual path printed above.
DATASET_DIR = Path("/content/drive/MyDrive/ourdataset/ourdataset")

NORMAL_DIR = DATASET_DIR / "train" / "normal"
ABNORMAL_DIR = DATASET_DIR / "train" / "abnormal"

def list_audio_files(folder):
    return [
        p for p in folder.rglob("*")
        if p.is_file() and p.suffix.lower() == ".wav"
    ]

normal_files = list_audio_files(NORMAL_DIR)
abnormal_files = list_audio_files(ABNORMAL_DIR)

print("Normal recordings:", len(normal_files))
print("Abnormal recordings:", len(abnormal_files))
print("Total recordings:", len(normal_files) + len(abnormal_files))
```
