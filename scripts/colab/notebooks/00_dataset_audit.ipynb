# CardioAgent-Ops — Dataset Audit

## Objective
Inspect and validate the heart-sound dataset before model training.

## Dataset
- Source: PASCAL and/or PhysioNet/CinC Challenge 2016
- Classes: Normal and Abnormal
- Format: WAV

## Audit tasks
1. Dataset discovery
2. Class distribution
3. Audio metadata
4. Duration distribution
5. Corrupted file detection
6. Duplicate detection
7. Dataset provenance verification

## Important
No training, segmentation, deletion, or modification
of original dataset files in this notebook.
